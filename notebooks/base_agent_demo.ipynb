# Base Agent Demonstration

This notebook walks through the core pieces of the `BaseAgent`.
It connects to the live OpenAI API (set `OPENAI_API_KEY`) and exercises a locally registered MCP tool.
Run the notebook inside the project environment after installing the package in editable mode (`pip install -e .[dev]`).


```python
from pathlib import Path

from project_agent_mcp import (
    AgentRunner,
    build_runner,
    register_add_numbers_tool,
)

print("Project components imported successfully.")
```

## Configure the OpenAI client
Set the `OPENAI_API_KEY` environment variable before running this cell.

```python
import os
model_name = "gpt-4o-mini"
if not os.environ.get("OPENAI_API_KEY"):
    raise RuntimeError("OPENAI_API_KEY environment variable must be set to use the live runner.")
print(f"Using OpenAI model '{model_name}'.")
```

```python
runner: AgentRunner = build_runner(
    model=model_name,
    tool_factories=[register_add_numbers_tool],
    dashboard_dir=Path("dashboards") / "notebook_demo",
    dashboard_filename="demo.html",
)
telemetry = runner.telemetry
print("BaseAgent runner ready.")
```

```python
response = runner.run_once("What is 2 plus 5?")
print("Agent response:", response)

dashboard_path = telemetry.dashboard_dir / telemetry.dashboard_filename
print("Dashboard generated at:", dashboard_path.resolve())
```
