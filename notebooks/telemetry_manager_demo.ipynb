# Telemetry Manager Demo

Record spans/events and render the standalone dashboard artifact for inspection.


```python
from pathlib import Path
from project_agent_mcp import TelemetryManager

telemetry = TelemetryManager(
    dashboard_dir=Path("dashboards") / "telemetry_demo",
    dashboard_filename="demo.html",
)

with telemetry.span("demo-span", {"step": 1}):
    telemetry.record_event("demo_event", {"message": "Hello telemetry"})

telemetry.write_dashboard()
```
