# MCP Client Demo

Register local tools with `MCPClient`, inspect the discovered schema, and execute a tool.


```python
from project_agent_mcp import MCPClient
client = MCPClient()

client.register_tool(
    name="add_numbers",
    handler=lambda a, b: a + b,
    description="Add two numbers.",
    input_schema={
        "type": "object",
        "properties": {
            "a": {"type": "number"},
            "b": {"type": "number"},
        },
        "required": ["a", "b"],
    },
)

client.list_tools()
```

```python
client.execute_tool("add_numbers", {"a": 2, "b": 5})
```
